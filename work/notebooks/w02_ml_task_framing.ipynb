## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

### My lane: Refresh / Content Opportunity Scoring

**Task type: Scoring**

The goal is to assign each content item a risk/opportunity score indicating how much it may need review or monitoring.

Scoring fits this lane because the output is intended to prioritize content rather than make an automatic yes/no decision. The score can combine multiple search-performance signals, such as previous impressions, query visibility, query concentration, and ranking-position volatility, into a useful review priority.

The output supports a content action: editors can review the highest-priority content first.

```python
# Basic imports and reproducibility check
import pandas as pd
import numpy as np

RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)

print("Environment ready.")
print("Random seed:", RANDOM_SEED)
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

### Target / Proxy

The score would estimate how much a content item may need review based on its likelihood of experiencing an impressions decline.

The underlying proxy is a defined outcome: a content item is considered to have declined when its latest-period impressions are below 80% of its previous-period impressions.

This is a rule-defined proxy rather than a directly observed business outcome. It is used to create a consistent target for the analysis while keeping the interpretation directional and decision-support focused.

```python
# Define the target/proxy rule used in this framing exercise

DECLINE_THRESHOLD = 0.80

print("Decline threshold:", DECLINE_THRESHOLD)
print("Target rule: latest-period impressions < 80% of previous-period impressions")
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

### Success Metric

The main success metric is **Precision@K**.

A good score means that a high proportion of the top-ranked content items are actually observed to experience an impressions decline in the evaluation period.

This metric matches the decision because the practical goal is to give editors a useful shortlist of content to review first, rather than classify every content item perfectly.

```python
# Check the success metric definition

K = 10

print("Success metric: Precision@K")
print("K =", K)
print("Interpretation: proportion of the top-K ranked items that are actually observed as declining.")
```

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

### Unit of Analysis

The unit of analysis is **one content item for a defined performance period**.

Each row represents one content item with aggregated search-performance signals calculated from its available history. The main fields describe previous impressions, query visibility, query concentration, and ranking-position volatility.

The practical output is therefore a content-level score that can be used to prioritize individual content items for review or monitoring.

```python
# Create a small example of the content-level analysis structure.
# One row = one content item for a defined performance period.

unit_df = pd.DataFrame({
    "content_id": ["content_001", "content_002", "content_003"],
    "previous_impressions": [1200, 850, 2400],
    "query_visibility": [0.62, 0.41, 0.78],
    "top_query_share": [0.35, 0.52, 0.29],
    "position_volatility": [1.8, 3.2, 1.1]
})

print("One row = one content item for a defined performance period.")
display(unit_df)
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

### Why ML Beats a Fixed Rule

A fixed rule can identify simple cases, but search-performance behavior depends on several signals at the same time.

For example, a content item may have declining impressions but also strong query visibility, while another may have stable impressions but high ranking-position volatility. A single threshold would not capture these different patterns well.

ML can learn how multiple signals relate to the defined decline proxy and produce a more useful prioritization score. The output is still used for human review and decision support rather than automatic content changes.

```python
# Simple check showing why multiple signals are useful
feature_columns = [
    "previous_impressions",
    "query_visibility",
    "top_query_share",
    "position_volatility"
]

print("Signals used for prioritization:")
for feature in feature_columns:
    print("-", feature)

print("\nNumber of signals:", len(feature_columns))
```
