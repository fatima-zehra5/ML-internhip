## 2. Data

*Which release, which tables, date windows, what you excluded and why. Public-safe.*

## Data

This study uses the full pseudonymized FlyRank internship warehouse release hosted on Hugging Face.

The analysis uses:
- `fact_content_daily_performance` for daily search-performance signals.
- `fact_content_query_90d` for query-level visibility and concentration signals.
- `dim_clients` to understand the available history for each client.

For the experiment, a 90-day analysis window was used with a minimum previous-period impression threshold of 200. Content items without enough history were excluded.

The analysis works with aggregated and anonymized data. No client names, domains, private queries, credentials, or raw exports are included in the public analysis.

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

## Methodology

The experiment uses aggregated content-level features rather than raw search records.

### Features
- Previous-period impressions
- Query visibility
- Rare and anonymized impression shares
- Top-query share
- Ranking-position volatility

### Label
A content item is classified as declining when its impressions in the latest period are below 80% of the previous period.

### Model
A Random Forest classifier is used as the predictive model. A majority-class baseline is used for comparison.

### Validation
Instead of a random row-level split, GroupShuffleSplit was used with `client_hash_id` as the group. This keeps clients separated between training and testing and gives a more realistic test of generalization to unseen clients.

### Leakage check
Features are calculated from the period available before the outcome being evaluated. Client grouping also prevents the same client's observations from being split across training and testing.

The experiment uses fixed random seeds for reproducibility. Results are interpreted as measured, directional evidence for decision support rather than causal evidence.

## 5. Limitations

*What this work cannot claim.*

## Limitations

- The dataset is pseudonymized, so the analysis cannot identify individual clients or domains.
- The results describe associations in the available search-performance data and do not establish causation.
- Performance may vary for clients or content outside the tested sample.
- The model was evaluated on 12 unseen clients, so the result should not be treated as a universal performance estimate.
- The analysis identifies content that may warrant review; it does not guarantee that a recommended action will improve search performance.
- Query-level data is aggregated and anonymized, so some detailed search behavior is not observable.

## 6. Ranked recommendations

*The action playbook output — the paper's recommendations section.*

## Ranked Recommendations

Based on the measured signals and model output, the following action priorities are recommended:

1. **Review high-risk declining content first** — prioritize content items that the model identifies as likely to decline.
2. **Investigate ranking-position volatility** — content with unstable average position should receive additional monitoring.
3. **Review query concentration** — highly concentrated query visibility may indicate dependence on a small number of search queries.
4. **Prioritize pages with sufficient historical impressions** — stronger historical evidence should be preferred over very low-volume content.
5. **Use the model as a review-prioritization tool** — recommendations should support human decisions rather than automatically trigger content changes.
